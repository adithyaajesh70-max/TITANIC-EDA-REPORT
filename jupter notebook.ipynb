# Task 5: Exploratory Data Analysis (EDA) – Titanic Dataset

**Objective:** Extract insights using visual and statistical exploration.

**Tools:** Python (Pandas, Matplotlib, Seaborn)

**Dataset:** Titanic passenger dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_theme(style='whitegrid')
df = pd.read_csv('titanic.csv')
df.head()
```

## 1. Dataset Inspection

```python
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
print('\nData information:')
df.info()
```

```python
df.describe(include='all').T
```

```python
print('Survived:\n', df['Survived'].value_counts())
print('\nSex:\n', df['Sex'].value_counts())
print('\nPassenger class:\n', df['Pclass'].value_counts())
print('\nEmbarked:\n', df['Embarked'].value_counts())
```

## 2. Missing-Value Analysis

```python
missing=pd.DataFrame({'Missing':df.isnull().sum(),'Percent':(df.isnull().mean()*100).round(2)}).sort_values('Missing',ascending=False)
missing[missing['Missing']>0]
```

```python
plt.figure(figsize=(10,5))
sns.heatmap(df.isnull(),cbar=False,yticklabels=False)
plt.title('Missing Values Heatmap')
plt.show()
```

**Observation:** Missing values are concentrated mainly in Cabin, Age and Embarked. Cabin has the largest proportion of missing values.

## 3. Univariate Analysis

```python
fig,ax=plt.subplots(1,3,figsize=(15,4))
sns.histplot(df['Age'].dropna(),bins=30,kde=True,ax=ax[0]); ax[0].set_title('Age Distribution')
sns.countplot(data=df,x='Sex',ax=ax[1]); ax[1].set_title('Passenger Count by Sex')
sns.countplot(data=df,x='Pclass',ax=ax[2]); ax[2].set_title('Passenger Count by Class')
plt.tight_layout(); plt.show()
```

**Observations:** Most passengers were young to middle-aged adults. Male passengers outnumbered female passengers, and third class was the largest passenger group.

```python
plt.figure(figsize=(9,5))
sns.boxplot(data=df,x='Pclass',y='Fare')
plt.title('Fare Distribution by Passenger Class')
plt.show()
```

**Observation:** First-class passengers generally paid much higher fares, with several high-value outliers.

## 4. Survival Analysis

```python
print(f"Overall survival rate: {df['Survived'].mean()*100:.2f}%")
print('\nSurvival rate by sex (%):')
print((df.groupby('Sex')['Survived'].mean()*100).round(2))
print('\nSurvival rate by class (%):')
print((df.groupby('Pclass')['Survived'].mean()*100).round(2))
```

```python
fig,ax=plt.subplots(1,2,figsize=(14,5))
sns.barplot(data=df,x='Sex',y='Survived',estimator='mean',errorbar=None,ax=ax[0]); ax[0].set_title('Survival Rate by Sex')
sns.barplot(data=df,x='Pclass',y='Survived',estimator='mean',errorbar=None,ax=ax[1]); ax[1].set_title('Survival Rate by Passenger Class')
plt.tight_layout(); plt.show()
```

**Observations:** Female passengers had a much higher survival rate than male passengers. Survival decreased from first class to third class.

```python
plt.figure(figsize=(9,5))
sns.boxplot(data=df,x='Survived',y='Age')
plt.title('Age Distribution by Survival')
plt.xlabel('Survived (0=No, 1=Yes)')
plt.show()
```

**Observation:** Age distributions overlap substantially, so age alone does not clearly separate survivors from non-survivors.

## 5. Relationship Analysis

```python
num=df.select_dtypes(include=np.number)
plt.figure(figsize=(10,7))
sns.heatmap(num.corr(),annot=True,cmap='coolwarm',fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

**Observation:** Passenger class and fare are strongly related. Survival is associated with higher class and fare.

```python
plt.figure(figsize=(9,6))
sns.scatterplot(data=df,x='Age',y='Fare',hue='Survived',alpha=.65)
plt.title('Age vs Fare by Survival')
plt.show()
```

**Observation:** Age and fare do not show a simple linear relationship. Fare varies widely across passengers.

```python
sns.pairplot(df[['Survived','Pclass','Age','SibSp','Parch','Fare']].dropna(),hue='Survived',corner=True)
plt.show()
```

## 6. Key Findings and Conclusion

- **891** passenger records and **12** variables were analyzed.
- Overall survival was **38.4%**.
- Female passengers had a much higher survival rate than male passengers.
- First-class passengers had the highest survival rate; third-class passengers had the lowest.
- Fare has high-value outliers and is associated with passenger class.
- Missing data is concentrated in Cabin, Age and Embarked.
- Sex and passenger class are the clearest patterns associated with survival.

**Conclusion:** EDA reveals important patterns before predictive modeling and highlights the need for careful missing-value and outlier handling.
